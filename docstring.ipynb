```python
import functools
def Dec(func):
    @functools.wraps(func)
    def Inner():
        func()
    return Inner
@Dec
def fun():
    """This is Docstring"""
    print("Hello")
print(fun.__doc__)
print(fun.__name__)
```

```python
import time
import functools


def timer(func):
    @functools.wraps(func)
    def wrapper(*args,**kwargs):
        start=time.asctime()
        start=time.time()
        print("Started:",start)
        print(f"Sum : {func(*args,**kwargs)}")
        end=time.asctime().split()
        end=time.time()
        print("End : {end}")
        print("Time.taken:",end-start)
    return wrapper
@timer
def add(x,y):
    """"This is Doc String"""
    su=0
    for i in range(1,x+y+1):
        su+=i
    return su
add(100000,200000)
print(add.__doc__)
print(time.asctime())
print(list(time.asctime().split()))
```
